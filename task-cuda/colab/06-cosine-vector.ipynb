```python
import pandas as pd
import matplotlib.pyplot as plt
path = "/tmp/06-cosine-vector.csv"
df = pd.read_csv(path)
size_df = df[df["mode"] == "size"]
block_df = df[df["mode"] == "block"]
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
if len(size_df):
    ax[0].plot(size_df["n"], size_df["ms"], marker="o")
    ax[0].set_xlabel("n")
    ax[0].set_ylabel("ms")
    ax[0].set_title("time vs size")
    ax[0].grid(True)
if len(block_df):
    ax[1].plot(block_df["block"], block_df["ms"], marker="o")
    ax[1].set_xlabel("block")
    ax[1].set_ylabel("ms")
    ax[1].set_title("time vs block")
    ax[1].grid(True)
plt.tight_layout()
plt.show()
```
